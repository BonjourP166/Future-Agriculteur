```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import glob
import pickle
import os

from statsmodels.tsa.statespace.sarimax import SARIMAX

# =========================================================
# CONFIG
# =========================================================
MODEL_DIR = "models_sarima_weekly"
os.makedirs(MODEL_DIR, exist_ok=True)

MIN_DATA_POINTS = 10  # minimum semaines


# =========================================================
# 1. CHARGEMENT CSV
# =========================================================
print("⏳ Chargement des fichiers SYNOP...")

files = sorted(glob.glob("synop_*.csv"))

if not files:
    raise FileNotFoundError("Aucun fichier synop_*.csv trouvé.")

cols = ["validity_time", "t", "name"]

df_raw = pd.concat(
    [pd.read_csv(f, sep=None, engine="python", usecols=cols) for f in files],
    ignore_index=True
)

df_raw["date"] = pd.to_datetime(df_raw["validity_time"], utc=True, errors="coerce")
df_raw["temperature"] = pd.to_numeric(df_raw["t"], errors="coerce") - 273.15

df_raw = df_raw.dropna(subset=["date", "temperature", "name"]).sort_values("date")

print(f"✅ Données chargées : {len(df_raw)} lignes")


# =========================================================
# 2. LISTE DES STATIONS
# =========================================================
stations = df_raw["name"].unique()

print(f"📊 Nombre de stations détectées : {len(stations)}")


# =========================================================
# 3. ENTRAÎNEMENT PAR STATION
# =========================================================
success = 0
skipped = 0

for station in stations:
    print(f"\n🔄 Station : {station}")

    df = df_raw[df_raw["name"] == station].copy()

    if df.empty:
        print("❌ Pas de données")
        skipped += 1
        continue

    # agrégation hebdomadaire
    weekly = df.groupby(pd.Grouper(key="date", freq="W-MON"))["temperature"].mean().dropna()

    if len(weekly) < MIN_DATA_POINTS:
        print("⚠️ Pas assez de données")
        skipped += 1
        continue

    try:
        model = SARIMAX(
            weekly,
            order=(1, 1, 1),
            seasonal_order=(0, 1, 1, 52),
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fit = model.fit(disp=False)

        # nom fichier safe
        station_safe = station.replace(" ", "_").replace("-", "_")

        model_path = os.path.join(
            MODEL_DIR,
            f"sarima_weekly_{station_safe}.pkl"
        )

        with open(model_path, "wb") as f:
            pickle.dump(fit, f)

        print(f"✅ Modèle sauvegardé : {model_path}")
        success += 1

    except Exception as e:
        print(f"❌ Erreur : {e}")
        skipped += 1


# =========================================================
# FIN
# =========================================================
print("\n🎯 TERMINÉ")
print(f"✅ Succès : {success}")
print(f"⚠️ Ignorées : {skipped}")
```

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import glob
import pickle
import os

from statsmodels.tsa.statespace.sarimax import SARIMAX

# =========================================================
# CONFIG
# =========================================================
MODEL_DIR = "models_sarima_monthly"
os.makedirs(MODEL_DIR, exist_ok=True)

MIN_DATA_POINTS = 5  # minimum mois


# =========================================================
# 1. CHARGEMENT CSV
# =========================================================
print("⏳ Chargement des fichiers SYNOP...")

files = sorted(glob.glob("synop_*.csv"))

if not files:
    raise FileNotFoundError("Aucun fichier synop_*.csv trouvé.")

cols = ["validity_time", "t", "name"]

df_raw = pd.concat(
    [pd.read_csv(f, sep=None, engine="python", usecols=cols) for f in files],
    ignore_index=True
)

df_raw["date"] = pd.to_datetime(df_raw["validity_time"], utc=True, errors="coerce")
df_raw["temperature"] = pd.to_numeric(df_raw["t"], errors="coerce") - 273.15

df_raw = df_raw.dropna(subset=["date", "temperature", "name"]).sort_values("date")

print(f"✅ Données chargées : {len(df_raw)} lignes")


# =========================================================
# 2. LISTE DES STATIONS
# =========================================================
stations = df_raw["name"].unique()
print(f"📊 Nombre de stations détectées : {len(stations)}")


# =========================================================
# 3. ENTRAÎNEMENT PAR STATION
# =========================================================
success = 0
skipped = 0

for station in stations:
    print(f"\n🔄 Station : {station}")

    df = df_raw[df_raw["name"] == station].copy()

    if df.empty:
        print("❌ Pas de données")
        skipped += 1
        continue

    # =====================================================
    # AGRÉGATION MENSUELLE
    # =====================================================
    monthly = df.groupby(pd.Grouper(key="date", freq="ME"))["temperature"].mean().dropna()

    if len(monthly) < MIN_DATA_POINTS:
        print("⚠️ Pas assez de données mensuelles")
        skipped += 1
        continue

    try:
        model = SARIMAX(
            monthly,
            order=(1, 1, 1),
            seasonal_order=(0, 1, 1, 12),  # saisonnalité annuelle
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fit = model.fit(disp=False)

        # sécuriser le nom
        station_safe = station.replace(" ", "_").replace("-", "_")

        model_path = os.path.join(
            MODEL_DIR,
            f"sarima_monthly_{station_safe}.pkl"
        )

        with open(model_path, "wb") as f:
            pickle.dump(fit, f)

        print(f"✅ Modèle sauvegardé : {model_path}")
        success += 1

    except Exception as e:
        print(f"❌ Erreur : {e}")
        skipped += 1


# =========================================================
# FIN
# =========================================================
print("\n🎯 TERMINÉ")
print(f"✅ Succès : {success}")
print(f"⚠️ Ignorées : {skipped}")
```
